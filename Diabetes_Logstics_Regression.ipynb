Data Preprocessing

```python
df.isnull().sum()
```

```python
df.loc[df['BMI'] == 0]
```

```python
df.BloodPressure.replace(0,np.median(df.BloodPressure),inplace=True)
df.SkinThickness.replace(0,np.median(df.SkinThickness),inplace=True)
df.Insulin.replace(0,np.median(df.Insulin),inplace=True)
df.BMI.replace(0,np.median(df.BMI),inplace=True)
df.Glucose.replace(0,np.median(df.Glucose),inplace=True)
```

```python
df
```

```python
from sklearn.preprocessing import MinMaxScaler
sc=MinMaxScaler()
dl=['Pregnancies','Outcome']
data1=sc.fit_transform(df.drop(dl,axis=1))
```

```python
con_data=df[["Pregnancies","Outcome"]]
```

```python
data2=pd.DataFrame(data1,columns=['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin',
       'BMI', 'DiabetesPedigreeFunction', 'Age'])
```

```python
df.columns
```

```python
final_df=pd.concat([con_data,data2],axis=1)
```

```python
final_df
```

```python
sns.heatmap(data2.corr(numeric_only=True),annot=True)
```

Model Creation


```python
x=final_df[['Pregnancies','Glucose','BloodPressure','SkinThickness','Insulin','BMI','DiabetesPedigreeFunction','Age']]
y=final_df.Outcome
```

```python
x
```

```python
print(x.columns)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=45)
```

```python
y
```

```python
y_test
```

```python
from sklearn.linear_model import LogisticRegression
LR=LogisticRegression()
LR.fit(x_train,y_train)
```

```python
y_pred=LR.predict(x_test)
```

```python
y_pred
```

```python
y_pred_prob=LR.predict_proba(x_test)
```

```python
y_pred_prob
```

```python
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    recall_score,
    precision_score,
    classification_report,
    f1_score
)
```

```python
cm=confusion_matrix(y_test,y_pred)
print(cm)
```

```python
recall=recall_score(y_test,y_pred)
print(recall)
```

```python
precision=precision_score(y_test,y_pred)
print(precision)
```

```python

f1score=f1_score(y_test,y_pred)
f1score
```
